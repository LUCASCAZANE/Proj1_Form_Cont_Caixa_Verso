# Introdução

Este projeto utiliza a base de operações de garantia do Sebrae/FAMPE, que reúne informações sobre solicitações de crédito feitas por empresas de diferentes portes, setores e estados.  
O objetivo é analisar o comportamento dessas operações, identificar padrões relevantes e responder perguntas de negócio que ajudem a entender o perfil das empresas atendidas e a dinâmica das garantias concedidas.

# Descrição da Base

A base utilizada neste projeto reúne informações de operações de crédito garantidas pelo Sebrae/FAMPE.  
Ela contém registros de empresas que solicitaram garantia para obter financiamento, incluindo dados sobre:

- características da empresa (porte, setor, faturamento)  
- localização (UF e município)  
- valores de crédito solicitados  
- datas das operações  
- tipo de garantia  
- status da solicitação  

Cada linha da base representa **uma operação de crédito**, permitindo analisar o comportamento das empresas atendidas, a distribuição dos valores financiados e a demanda por garantias ao longo do tempo.

# Perguntas de Negócio

1. **Qual é o perfil das empresas que mais solicitam garantia?**  
2. **Como evoluiu o volume de operações ao longo dos anos?**  
3. **Quais setores econômicos concentram os maiores valores de crédito?**  
4. **Existe relação entre o valor solicitado e o porte da empresa?**  
5. **Quais estados apresentam maior demanda por garantia?**
