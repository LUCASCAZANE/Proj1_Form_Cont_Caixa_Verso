# Descrição da Base

A base utilizada neste projeto reúne informações de operações de crédito garantidas pelo Sebrae/FAMPE.  
Ela contém registros de empresas que solicitaram garantia para obter financiamento, incluindo dados sobre:

- características da empresa (porte, setor, faturamento)  
- localização (UF e município)  
- valores de crédito solicitados  
- datas das operações  
- tipo de garantia  
- status da solicitação  

Cada linha da base representa **uma operação de crédito**, permitindo analisar o comportamento das empresas atendidas, a distribuição dos valores financiados e a demanda por garantias ao longo do tempo.
